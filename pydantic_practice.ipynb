```python
from dataclasses import dataclass
@dataclass
class Person():
  name:str
  age:int
  city:str

person = Person(name="melika", age=27, city=12)
print(person)
```

```python
class Person(BaseModel):
  name:str
  age:int
  city:str

person = Person(name="melika", age=27, city="Rasht")
print(person)
```

```python
#Model with optional fields
from typing import Optional
class Employee(BaseModel):
  id:int
  name:str
  department:str
  salary:Optional[float] = None
  is_active:Optional[bool] = True
```

```python
emp1 = Employee(id=1, name="john", department= "IT")
print(emp1)
```

```python
from pydantic import BaseModel
from typing import List
class Classroom(BaseModel):
  room_number:str
  students:List[str]
  capacity:int
```

```python
classroom = Classroom(
    room_number='A101',
    students=['Alice', 'Bob', 'Sara'],
    capacity=30
)
print(classroom)
```

```python
try:
  invalid_val = Classroom(room_number='101', students= ['sara',123], capacity=30)
except ValueError as e:
  print(e)
```

```python
##Model with nested models
from pydantic import BaseModel

class Address(BaseModel):
  street:str
  city:str
  zip_code:str
class Customer(BaseModel):
  customer_id: int
  name: str
  address: Address
customer = Customer(
    customer_id = 1,
    name = "melika",
    address = {"street":"123 main street", "city": "Rasht", "zip_code": "12034"}

)
print(customer)
```

```python
#pydantic fields: Customization and constraints
from pydantic import BaseModel, Field
class Item(BaseModel):
  name:str=Field(min_length = 2, max_length = 50)
  price:float=Field(gt=0, le=1000)
  quantity:int=Field(ge=0)

item = Item(name="Book", price=29.99, quantity=20)
print(item)
```
